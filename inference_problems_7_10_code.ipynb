# Problems 7–10 — Code Only

```python
import numpy as np
import pandas as pd

critical_values = {0.90: 1.645, 0.95: 1.96, 0.99: 2.576}
```

## Problem 7 — Central air proportion

```python
df = pd.read_csv('' \
'data/ames_prices.csv')

central_air = (df['Central.Air'] == 'Y').astype(int)
p_hat = central_air.mean()

rng = np.random.default_rng(42)
R = 2000
results = []
for t in range(R):
    idx = rng.integers(0, len(central_air), len(central_air))
    p_hat_t = central_air.iloc[idx].mean()
    results.append(p_hat_t)
results = np.array(results)

standard_error = np.std(results, ddof=1)
print(f'Seed: 42, bootstrap replications: {R}')
print(f'Estimated proportion: {p_hat:.3f}')
print(f'Bootstrap standard error: {standard_error:.3f}')

for confidence, c in critical_values.items():
    lower = p_hat - c * standard_error
    upper = p_hat + c * standard_error
    decision = 'Fail to reject' if lower <= 0.90 <= upper else 'Reject'
    print(f'{confidence:.0%} CI: ({lower:.3f}, {upper:.3f}); {decision} H0: p = 0.90')
```

## Problem 8 — Correlation of transformed area and price

```python
df = pd.read_csv('data/ames_prices.csv')
rdf = df[['area', 'price']].dropna()

x = np.arcsinh(rdf['area'])
y = np.arcsinh(rdf['price'])
r_hat = np.corrcoef(x, y)[0, 1]

rng = np.random.default_rng(42)
R = 2000
results = []
for t in range(R):
    idx = rng.integers(0, len(rdf), len(rdf))
    r_hat_t = np.corrcoef(x.iloc[idx], y.iloc[idx])[0, 1]
    results.append(r_hat_t)
results = np.array(results)

standard_error = np.std(results, ddof=1)
print(f'Seed: 42, bootstrap replications: {R}')
print(f'Estimated correlation: {r_hat:.3f}')
print(f'Bootstrap standard error: {standard_error:.3f}')

for confidence, c in critical_values.items():
    lower = r_hat - c * standard_error
    upper = r_hat + c * standard_error
    decision = 'Fail to reject' if lower <= 0 <= upper else 'Reject'
    print(f'{confidence:.0%} CI: ({lower:.3f}, {upper:.3f}); {decision} H0: rho = 0')
```

## Problem 9 — METABRIC recorded survival/follow-up time

```python
df = pd.read_csv('data/metabric.csv')
survival = df['Overall Survival (Months)'].dropna()
mean_hat = survival.mean()

rng = np.random.default_rng(42)
R = 2000
results = []
for t in range(R):
    idx = rng.integers(0, len(survival), len(survival))
    mean_hat_t = survival.iloc[idx].mean()
    results.append(mean_hat_t)
results = np.array(results)

standard_error = np.std(results, ddof=1)
print(f'Seed: 42, bootstrap replications: {R}')
print(f'Estimated mean: {mean_hat:.3f} months ({mean_hat / 12:.3f} years)')
print(f'Bootstrap standard error: {standard_error:.3f} months')

for confidence, c in critical_values.items():
    lower = mean_hat - c * standard_error
    upper = mean_hat + c * standard_error
    decision = 'Fail to reject' if lower <= 120 <= upper else 'Reject'
    print(f'{confidence:.0%} CI: ({lower:.3f}, {upper:.3f}) months')
    print(f'         ({lower / 12:.3f}, {upper / 12:.3f}) years; {decision} H0: mu = 120')
```

## Problem 10 — NHANES weekday sleep for adults

```python
df = pd.read_csv('data/nhanes_data_17_18.csv')
print(df.columns.tolist())
```
